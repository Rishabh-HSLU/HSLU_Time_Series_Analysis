### Visualization
#### Seasonal plot
 - weekly pattern behavior(speculation, checking through a seasonal plot). We select the jan month of the time series

```python
df1 = df1.resample('h').mean()
df1_jan_month = df1.iloc[1:720, :]
fig = px.line(df1_jan_month, title='Basel Energy Demand in Kwh  in the month of JANUARY')
fig.show()
```

```python
df = df1_jan_month.copy()

df['DayOfWeek'] = df.index.dayofweek
df['Week'] = df.index.isocalendar().week

pivot_df = df.pivot_table(index='DayOfWeek', columns='Week',
                           values='consumptions(kWh)', aggfunc='mean')
pivot_df.dropna(axis = 1, inplace = True)


# Plot each week
fig = go.Figure()
day_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

for week in pivot_df.columns:
    fig.add_trace(go.Scatter(
        x=day_names,
        y=pivot_df[week],
        mode='lines+markers',
        name=f'Week {week}'
    ))

fig.update_layout(
    title='Seasonal Plot: Weekly Consumption Pattern',
    xaxis_title='Day of Week',
    yaxis_title='Consumption (kWh)',
    hovermode='x unified'
)

fig.show()
```

## Factors Influencing the data
- seasons(e.g. winter and summer) affect the magnitude of consumption levels.
- The downward trend in the consumption magnitude could be because the energy efficiency through electrical grid got better

## Expected patterns
- Downward trend in energy cosumption
- seasonal behavior with multiple frequencies: day/night, weekly, yearly
